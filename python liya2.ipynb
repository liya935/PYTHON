```python
names = input("Enter first names separated by spaces: ").split()

count = 0

for name in names:
    count += name.lower().count('a')

print("Names:", names)
print("Total occurrences of 'a':", count)
```

```python
color_list1 = input("enter the colors for list 1: ").split()
color_list2 = input("enter the colors for list 2: ").split()

for color in color_list1:
    if color not in color_list2:
        print(color)
```

```python
d ={}

n = int(input("enter the number of items:"))

for i in range(n):
    key = input("enter key: ")
    value = int(input("enter value: "))
    d[key] =  value
    
    print("ascending order:",dict(sorted(d.items())))
    print("descending order:",dict(sorted(d.items(),reverse=True)))
```

```python
d1 = {}
d2 = {}

n = int(input("Enter number of items in first dictionary: "))
for i in range(n):
    key = input("Enter key: ")
    value = input("Enter value: ")
    d1[key] = value

n = int(input("Enter number of items in second dictionary: "))
for i in range(n):
    key = input("Enter key: ")
    value = input("Enter value: ")
    d2[key] = value

d1.update(d2)

print("Merged dictionary:", d1)
```
