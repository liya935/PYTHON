```python

list1 = list(map(int, input("Enter first list of integers: ").split()))
list2 = list(map(int, input("Enter second list of integers: ").split()))

if len(list1) == len(list2):
    print("Both lists have the same length")
else:
    print("Both lists have different lengths")

if sum(list1) == sum(list2):
    print("Both lists have the same sum")
else:
    print("Both lists have different sums")

common = set(list1) & set(list2)

if common:
    print("Common values:", common)
else:
    print("No common values")
```

```python
numbers = list(map(int, input("Enter a list of integers: ").split()))

for i in range(len(numbers)):
    if numbers[i] > 100:
        numbers[i] = "over"

print("Updated list:", numbers)
```
